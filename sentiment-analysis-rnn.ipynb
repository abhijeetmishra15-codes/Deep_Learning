```python
import tensorflow
from tensorflow.keras.preprocessing.text import one_hot
```

```python
docs = ['the glass of milk',
       'my name is abhijeet',
       'i am a good boy',
       'i am a good developer',
       'the cup of tea']
```

```python
#define the vocabulary size

voc_size = 10000
```

```python
#One hot representation for every word

one_hot_rep = [one_hot(words , voc_size)for words in docs]
```

```python
one_hot_rep
```

```python
#word embedding representation

from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_length = 8

embedded_docs = pad_sequences(one_hot_rep , padding = 'pre' , maxlen = sent_length)
embedded_docs
```

```python
#Feature representation

dim = 10
```

```python
model = Sequential()

model.add(Embedding(voc_size , dim , input_length = sent_length))

model.compile(optimizer = 'adam' , loss='mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```

**Deep Learning Project Using RNN**

```python
import tensorflow as tf
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding , SimpleRNN , Dense
from tensorflow.keras.datasets import imdb
```

```python
max_features = 10000
(X_train , y_train) , (X_test , y_test) = imdb.load_data(num_words = max_features)
```

```python
print(X_train.shape , X_test.shape)
```

```python
sample_review = X_train[0]
sample_label = y_train[0]

print(f"sample review as integer values : {sample_review}")
print(f"sample label :{sample_label}")
```

```python
#mapping of word index back to words

word_index = imdb.get_word_index()

reverse_word_index = {value : key for key , value in word_index.items()}


decoded_review = ' '.join([reverse_word_index.get(i - 3 , '?') for i in sample_review])
decoded_review
```

```python
#padding Sequences

max_len = 500

X_train  = sequence.pad_sequences(X_train , maxlen = max_len)
X_test = sequence.pad_sequences(X_test , maxlen = max_len)
```

```python
#Training Simple RNN

model = Sequential()

model.add(Embedding(max_features, 128 , input_length = max_len))
model.add(SimpleRNN(128 , activation = 'relu'))
model.add(Dense(1 , activation = 'sigmoid'))
```

```python
model.summary()
```

```python
#Creating an instance of EarlyStopping Callback

from tensorflow.keras.callbacks import EarlyStopping

early_stopping = EarlyStopping(monitor='val_loss' , patience = 5 , restore_best_weights = True)
```

```python
model.compile(optimizer='adam' , loss='binary_crossentropy' , metrics = ['accuracy'])
history = model.fit(X_train, y_train , epochs = 20 , batch_size=32 , validation_split=0.2 , callbacks=[early_stopping])
```

```python
def decode_review(encoded_review):
    return ' '.join([reverse_word_index.get(i - 3 , '?') for i in encoded_review])


def preprocess_text(text):

    words = text.lower().split()
    encoded_review = [word_index.get(word, 2) + 3 for word in words]
    padded_review = sequence.pad_sequences([encoded_review] , maxlen = 500)

    return padded_review
```

```python
def predict_sentiment(review):
    preprocessed_input = preprocess_text(review)

    prediction = model.predict(preprocessed_input)

    sentiment = 'Positive' if prediction[0][0] > 0.5 else 'Negative'
    return sentiment , prediction[0][0]
```

```python
review = 'This movie was fantastic! The acting was great and the plot was thrilling.'

sentiment , score = predict_sentiment(review)
```

```python
print('Review :' , review)
print('Sentiment :' , sentiment)
print('Prediction Score :' , score)
```
