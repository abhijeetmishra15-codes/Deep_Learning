```python
import numpy as np
```

```python
docs = ['go india',
        'india india',
        'jeetega bhai jeetega india jeetega',
        'bharat mata ki jai',
        'kohli kohli',
        'sachin sachin',
        'dhoni dhoni',
        'inquilab zindabad']
```

```python
#tokenize input data

!pip install tensorflow

from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer(oov_token='abhijeet')
```

```python
tokenizer.fit_on_texts(docs)
```

```python
tokenizer.word_index #dictionary and alloted tokens
```

```python
tokenizer.word_counts
```

```python
tokenizer.document_count
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences

sequences =pad_sequences(sequences , padding='post')#apply post padding
```

```python
sequences
```

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train , y_train) , (X_test , y_test) = imdb.load_data()
```

```python
X_train.shape , X_test.shape
```

```python
print(len(X_train[0]))
print(len(X_train[1]))
```

```python
#Apply padding

X_train = pad_sequences(X_train , padding='post' , maxlen=1000)
X_test = pad_sequences(X_test , padding='post' , maxlen=1000)
```

```python
print(len(X_train[0]))
```

```python
model = Sequential()

model.add(SimpleRNN(32 , input_shape=(100 , 1) , return_sequences=False))
model.add(Dense(1 , activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy' , optimizer='adam' , metrics=['accuracy'])

model.fit(X_train , y_train , epochs=5 , validation_data=(X_test , y_test))
```

```python
#Embeddings

model = Sequential()

model.add(Embedding(17 , output_dim=2 , input_length=5))

model.summary()
```

```python
model.compile('adam' , 'accuracy')
```

```python
pred = model.predict(sequences)

print(pred)
```
